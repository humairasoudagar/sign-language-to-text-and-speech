```python
from pathlib import Path  # for working with local folders and files
import cv2
import matplotlib.pyplot as plt
import random
from PIL import Image  # opens images 
import numpy as np
```

```python
project_root = Path.cwd().parent  # main project folder
dataset_path = project_root / "datasets" / "asl_alphabet_train_set"  # main project folder to dataset folder
class_folders =[folder for folder in dataset_path.iterdir() if folder.is_dir()]  # checks all directory
print(len(class_folders))

print(dataset_path.exists())


```

```python
images = []
labels = []

for folder in class_folders:
    image = list(folder.glob("*.jpg"))
    selected_image = random.choice(image)

    img = Image.open(selected_image)

    images.append(img)
    labels.append(folder.name)
```

```python
fig, axes = plt.subplots(5,6, figsize=(15,12))  # 5 rows, 6 col, figsize is width and height of fig in inches 
for ax, img, label in zip(axes.flatten(), images, labels): 
    ax.imshow(img)
    ax.axis("off")
    ax.set_title(label)

for ax in axes.flatten()[len(images):]:
    ax.axis("off")

plt.tight_layout()  # spacing
plt.show()
```
